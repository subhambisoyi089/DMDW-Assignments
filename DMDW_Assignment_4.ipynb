## 1. Nominal attribute dissimilarity (as shown in the assignment)

```python
import pandas as pd
import numpy as np

data = pd.DataFrame({
    "Colour": ["Red", "Blue", "Red", "Green"],
    "City": ["Delhi", "Mumbai", "Delhi", "Kolkata"]
})
print("Nominal dataset:")
print(data)

n = len(data)
matrix = np.zeros((n, n))
for i in range(n):
    for j in range(n):
        matches = sum(data.iloc[i, col] == data.iloc[j, col]
                      for col in range(data.shape[1]))
        matrix[i, j] = 1 - matches / data.shape[1]

labels = [f"O{i+1}" for i in range(n)]
print("\\nNominal dissimilarity matrix:")
print(pd.DataFrame(matrix, index=labels, columns=labels).round(2))
```

## 2. Ordinal attribute dissimilarity (standard implementation)

```python
# Convert ordered categories to ranks, normalize to [0, 1],
# then calculate absolute pairwise differences.
education = ["Low", "Medium", "High", "Medium"]
rank_map = {"Low": 1, "Medium": 2, "High": 3}
ranks = np.array([rank_map[x] for x in education], dtype=float)
normalized = (ranks - 1) / (len(rank_map) - 1)
ordinal_matrix = np.abs(normalized[:, None] - normalized[None, :])

print("Ordinal values:", education)
print(pd.DataFrame(ordinal_matrix).round(2))
```

## 3. Binary attribute dissimilarity (standard implementation)

```python
# Simple matching dissimilarity for symmetric binary attributes:
# number of mismatches / number of attributes.
binary = np.array([
    [1, 0, 1],
    [1, 1, 0],
    [0, 0, 1],
    [1, 0, 1]
])
binary_matrix = np.zeros((len(binary), len(binary)))
for i in range(len(binary)):
    for j in range(len(binary)):
        binary_matrix[i, j] = np.mean(binary[i] != binary[j])

print("Binary data:")
print(binary)
print("\\nBinary dissimilarity matrix:")
print(pd.DataFrame(binary_matrix).round(2))
```
