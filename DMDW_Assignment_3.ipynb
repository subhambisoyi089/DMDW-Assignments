## 1. Create a sample student dataset

```python
import pandas as pd
import matplotlib.pyplot as plt

data = {
    "RollNo": list(range(101, 115)),
    "Age": [19, 20, 18, 20, 19, 21, 18, 22, 20, 19, 21, 25, 18, 23],
    "Gender": ["M", "M", "F", "M", "F", "M", "F", "M", "M", "F", "M", "F", "M", "F"],
    "Branch": ["CSE", "AIML", "CSE", "AIML", "ME", "CSE", "ME", "CSE", "AIML", "ME", "CSE", "ME", "AIML", "ME"],
    "Section": ["A", "B", "A", "B", "C", "A", "C", "B", "A", "C", "A", "C", "B", "C"]
}
df = pd.DataFrame(data)
df.to_csv("student_data.csv", index=False)
df.head()
```

## 2. Read a CSV file and display first five rows

```python
df = pd.read_csv("student_data.csv")
print(df)
print("\\nFirst five rows:")
print(df.head())
```

## 3. Display DataFrame information, columns, and data types

```python
print("Columns:", df.columns)
print("\\nData types:")
print(df.dtypes)
print("\\nDataFrame information:")
df.info()
```

## 4. Check missing values

```python
print(df.isnull())
print("\\nMissing values in each column:")
print(df.isnull().sum())
```

## 5. Select rows using iloc and loc

```python
print("First row using iloc:")
print(df.iloc[0])

print("\\nRow with index 0 using loc:")
print(df.loc[0])
```

## 6. Display students whose age is greater than 20

```python
print(df[df["Age"] > 20])
```

## 7. Display students whose age is between 19 and 20

```python
students_19_20 = df[df["Age"].between(19, 20)]
print(students_19_20)
```

## 8. Display all male candidates

```python
male_students = df[df["Gender"] == "M"]
print(male_students)
```

## 9. Sort students by age in ascending order

```python
sorted_students = df.sort_values(by="Age", ascending=True)
print(sorted_students)
```

## 10. Visualize age distribution

```python
plt.hist(df["Age"], bins=6, edgecolor="black")
plt.xlabel("Age")
plt.ylabel("Number of students")
plt.title("Student Age Distribution")
plt.show()
```
