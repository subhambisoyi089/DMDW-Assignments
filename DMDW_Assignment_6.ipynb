## 1. Import libraries and create the dataset

```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import MinMaxScaler, StandardScaler

data = pd.DataFrame({
    "age": [15, 16, 17, 18, 19, 20],
    "studytime": [1, 2, 3, 2, 4, 3],
    "failures": [0, 1, 0, 2, 1, 0]
})
print("Original data:")
print(data)
```

## 2. Min-Max normalization

```python
minmax_scaler = MinMaxScaler()
minmax_values = minmax_scaler.fit_transform(data)
minmax_df = pd.DataFrame(minmax_values, columns=data.columns)
print("Min-Max normalization:")
print(minmax_df)
```

## 3. Z-score normalization

```python
zscore_scaler = StandardScaler()
zscore_values = zscore_scaler.fit_transform(data)
zscore_df = pd.DataFrame(zscore_values, columns=data.columns)
print("Z-score normalization:")
print(zscore_df)
```

## 4. Decimal scaling

```python
# For each column, choose the smallest j such that max(abs(x))/10**j < 1.
decimal_scaled = pd.DataFrame(index=data.index)
for column in data.columns:
    maximum = data[column].abs().max()
    j = 0
    while maximum / (10 ** j) >= 1:
        j += 1
    decimal_scaled[column] = data[column] / (10 ** j)

print("Decimal scaling:")
print(decimal_scaled)
```

> Min-Max maps values to [0, 1]. Z-score uses mean 0 and standard deviation 1. Decimal scaling shifts the decimal point by dividing each column by a power of 10.
