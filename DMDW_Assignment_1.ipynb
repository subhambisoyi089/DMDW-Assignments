```python
student_marks = {"Apple": 1, "Banana": 2, "Orange": 3}
print(student_marks)
print(type(student_marks))
for key, value in student_marks.items():
    print(key, ":", value)
```

## 5. Find the greatest among three numbers

```python
def greatest(a, b, c):
    if a > b and a > c:
        return a
    elif b > a and b > c:
        return b
    elif c > a and c > b:
        return c
    return "There is a tie (two or more values are equal and greatest)"

a = int(input("Enter value for a: "))
b = int(input("Enter value for b: "))
c = int(input("Enter value for c: "))
print("Greatest:", greatest(a, b, c))
```

## 6. Check whether a number is prime

```python
def is_prime(n):
    if n < 2:
        return False
    for i in range(2, int(n ** 0.5) + 1):
        if n % i == 0:
            return False
    return True

number = int(input("Enter a number: "))
print(number, "is a prime number" if is_prime(number) else "is not a prime number")
```

## 7. Display Fibonacci series

```python
def fibonacci(count):
    a, b = 0, 1
    for _ in range(count):
        print(a)
        a, b = b, a + b

count = int(input("Enter the count: "))
fibonacci(count)
```

## 8. Copy content from one file to another
