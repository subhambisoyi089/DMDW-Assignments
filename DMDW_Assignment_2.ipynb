## Dataset

```python
data = [10, 12, 15, 18, 20, 20, 22, 25]
print("Dataset:", data)
```

## 1. Mean using logic and NumPy

```python
mean_logic = sum(data) / len(data)
print("Mean using logic:", mean_logic)

import numpy as np
print("Mean using NumPy:", np.mean(data))
```

## 2. Median using logic and NumPy

```python
sorted_data = sorted(data)
n = len(sorted_data)
if n % 2 == 0:
    median_logic = (sorted_data[n//2 - 1] + sorted_data[n//2]) / 2
else:
    median_logic = sorted_data[n//2]
print("Median using logic:", median_logic)
print("Median using NumPy:", np.median(data))
```

## 3. Mode using logic and statistics

```python
from collections import Counter
frequency = Counter(data)
max_frequency = max(frequency.values())
modes = [value for value, count in frequency.items() if count == max_frequency]
print("Mode(s) using logic:", modes)

import statistics
print("Mode using statistics:", statistics.mode(data))
```

## 4. Variance using logic and NumPy

```python
mean_value = sum(data) / len(data)
population_variance = sum((x - mean_value) ** 2 for x in data) / len(data)
sample_variance = sum((x - mean_value) ** 2 for x in data) / (len(data) - 1)
print("Population variance using logic:", population_variance)
print("Sample variance using logic:", sample_variance)
print("Population variance using NumPy:", np.var(data))
print("Sample variance using NumPy:", np.var(data, ddof=1))
```

## 5. Standard deviation using logic and NumPy

```python
import math
std_logic = math.sqrt(population_variance)
print("Standard deviation using logic:", std_logic)
print("Standard deviation using NumPy:", np.std(data))
```

> Note: NumPy's `var()` and `std()` use population formulas by default (`ddof=0`). Use `ddof=1` for sample statistics.
