## 1. Import libraries and create demonstration data

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.DataFrame({
    "traveltime": [1, 2, 1, 3, 2, 1, 4, 2, 3, 1],
    "studytime":  [3, 2, 4, 1, 2, 4, 1, 3, 2, 4]
})
print(df.head())
```

## 2. Visualize travel time vs study time

```python
sns.lineplot(data=df, x="traveltime", y="studytime", marker="o")
plt.title("Travel Time vs Study Time")
plt.xlabel("Travel Time")
plt.ylabel("Study Time")
plt.show()
```

## 3. Calculate correlation

```python
correlation = df["traveltime"].corr(df["studytime"])
print("Correlation coefficient:")
print(correlation)

correlation_matrix = df[["traveltime", "studytime"]].corr()
sns.heatmap(correlation_matrix, annot=True, cmap="YlGnBu", vmin=-1, vmax=1)
plt.title("Correlation Matrix")
plt.show()
```

## 4. Calculate covariance

```python
covariance = df[["traveltime", "studytime"]].cov()
print("Covariance matrix:")
print(covariance)

sns.heatmap(covariance, annot=True, cmap="YlGnBu")
plt.title("Covariance Matrix")
plt.show()
```
